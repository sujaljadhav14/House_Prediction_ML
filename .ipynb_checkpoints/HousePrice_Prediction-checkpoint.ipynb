```python
data["bath"] = data["bath"].fillna(data["bath"].median())
```

```python
data["bhk"] = data['size'].str.split().str.get(0).astype(int)
```

```python
data[data.bhk<20]
```

```python
data["total_sqft"].unique()
```

```python
def convertRange(x):
    temp = x.split("-")
    if len(temp) == 2:
        return(float(temp[0]) + float(temp[1]))/2
    try:
        return float(x)
    except:
        return None
```

```python
data["total_sqft"]=data["total_sqft"].apply(convertRange)
```

```python
data.head()
```

## price per square feet

```python
data['price_per_feet'] = data['price']*100000 / data['total_sqft']
```

```python
data["location"] = data["location"].apply(lambda x : x.strip())
location_count = data['location'].value_counts()
```

```python
location_count
```

```python
location_lessthan_10 = location_count[location_count<=10]
location_lessthan_10
```

```python
data['location']=data['location'].apply(lambda x: "other" if x in location_lessthan_10 else x)
```
